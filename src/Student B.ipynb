```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

df = pd.read_csv("GlobalLandTemperaturesByCountry_clean.csv")

df.head()
```

```python
thailand = df[df["Country"] == "Thailand"]["AverageTemperature"].dropna()
japan = df[df["Country"] == "Japan"]["AverageTemperature"].dropna()

t_stat, p_value = stats.ttest_ind(thailand, japan, equal_var=False)

print("Thailand mean:", thailand.mean())
print("Japan mean:", japan.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```

```python
thai = df[df["Country"] == "Thailand"]["AverageTemperature"].dropna()
japan = df[df["Country"] == "Japan"]["AverageTemperature"].dropna()
australia = df[df["Country"] == "Australia"]["AverageTemperature"].dropna()

f_stat, p_value = stats.f_oneway(thai, japan, australia)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference among countries")
else:
    print("No significant difference among countries")
```

```python
data = df["AverageTemperature"].dropna()

distributions = {
    "Normal": stats.norm,
    "Exponential": stats.expon,
    "Gamma": stats.gamma,
    "Lognormal": stats.lognorm,
    "Uniform": stats.uniform
}

for name, dist in distributions.items():
    params = dist.fit(data)
    ks_stat, p_value = stats.kstest(data, dist.name, args=params)

    print(name)
    print("KS statistic:", ks_stat)
    print("P-value:", p_value)
    print()
```

```python
data = df["AverageTemperature"].dropna()

n = len(data)
mean = data.mean()
std = data.std()

se = std / np.sqrt(n)

if n >= 30:
    z = stats.norm.ppf(0.975)
    margin = z * se
else:
    t = stats.t.ppf(0.975, df=n-1)
    margin = t * se

lower = mean - margin
upper = mean + margin

print("Number of data:", n)
print("Mean:", mean)
print("95% Confidence Interval:")
print("Lower:", lower)
print("Upper:", upper)
```

```python
data = df["AverageTemperature"].dropna().values

bootstrap_means = []

for i in range(10000):
    sample = np.random.choice(data, size=len(data), replace=True)
    bootstrap_means.append(np.mean(sample))

lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)

print("Original mean:", np.mean(data))
print("Bootstrap mean:", np.mean(bootstrap_means))
print("95% Bootstrap CI:")
print("Lower:", lower)
print("Upper:", upper)
```
